# File Handling in Python

##### Types of data used for I/O:
- Text - '12345' as a sequence of unicode chars
- Binary - 12345 as a sequence of bytes of its binary equivalent

##### Hence there are 2 file types to deal with
- Text files - All program files are text files
- Binary Files - Images,music,video,exe files



### How File I/O is done in most programming languages

- Open a file
- Read/Write data
- Close the file

# Writing to a file

```python
#if the file is not present 

f = open("files/sample.txt", "w") #open() function returns a file object, we store the object reference it to further use
f.write("Hello Python") #write() method writes the string to the file
f.close() #close() method closes the file object and saves the changes
```

```python
f.write("Hello") #we can not write to the file after closing it, it will give an error
```

```python
# write multiline strings
f = open('files/sample1.txt','w')
f.write('hello world')
f.write('\nhow are you?')
f.write("\nwho are you?")
f.close()
```

```python
# case 2 - if the file is already present write() method will overwrite the existing content of the file
f = open('files/sample1.txt','w')
f.write('Yo yo honey singh')
f.close()
```

```python
# introducing append mode which allows us to add content to the existing file without overwriting it
f = open('files/sample1.txt','a')
f.write('\nI am Tawfica')
f.close()
```

```python
# write lines
L = ['hello\n','hi\n','how are you\n','I am fine']

f = open('files/sample.txt','w')
f.writelines(L) # writelines() method writes a list of strings to the file
f.close()
```

# reading from files

```python
f = open('files/sample.txt','r') # open the file in read mode
s = f.read(15) # read() method reads the entire content of the file
print(s)
f.close()
```

```python
# reading upto n chars
f = open('files/sample.txt','r')
s = f.read(25)
print(s)
f.close()
```

```python
f = open('files/sample.txt','r')
data = f.readline() # readline() method reads a single line from the file
data = f.readline()
data = f.readline()
print(data)
f.close()
```

```python
# reading all lines
f = open("files/sample.txt", "r")
while True:
    data = f.readline()
    if data == "":
        break
    else:
        print(data, end= "")

f.close()
```

# File Handling using Context Manager (With)

### Using Context Manager (With)

- It's a good idea to close a file after usage as it will free up the resources
- If we dont close it, garbage collector would close it
- with keyword closes the file as soon as the usage is over

```python
#doing this we don't have to explicitly close the file, it will automatically close the file after the block of code is executed
with open('files/sample.txt', 'w') as f:
    f.write("Hello there!") 
```

```python
f.write("Tawfica") #we can not write to the file after closing it, it will give an error
```

```python
# read()
with open("files/sample.txt", "r") as f:
    s = f.read()
    print(s)
```

```python
# readline()
with open("files/sample.txt", "r") as f:
    s = f.readline(5)
    print(s)
```

```python
# append mode
# with open("files/sample.txt", "a") as f:
#     s = f.write("\nMy name is Bappy")
with open('files/sample.txt','a') as f:
    f.write("\nMyself Tawfica")
```

```python
# moving within a file
with open("files/sample.txt", "r") as f:
    print(f.read(20))
```

# Problems with working in text mode

- can't work with binary files like images
- not good for other data types like int/float/list/tuples

```python
# pyhton fails to decode binary files into text files
with open("files/iit.png", 'r') as f:
    print(f.read())
```

```python
# we can open the file in read-binary mode but we can not read the content of the file as it is in binary format
with open("files/iit.png", 'rb') as f:
    print(f.read())
```

```python
# if we want to see the content of a binary file to another binary file, we can do it by reading the content of the source file in binary mode and writing it to the destination file in binary mode
with open("files/iit.png", 'rb') as f:
    image_data =f.read()
    print(image_data)

with open("files/iit_copy.png", 'wb') as f1:
        f1.write(image_data)
```

### Serialization and Deserialization

- **Serialization** - process of converting python data types/object to JSON format to save or exchange 
- **Deserialization** - process of converting JSON to python data types

#### What is JSON?
The full form of JSON is JavaScript Object Notation.
- Lightweight Data Format
- we can store structured data using it
- we can exchange it by applications to applications
image.png

```python
#json module allows us to work with json data in python. It provides methods to convert python objects to json and vice versa
import json
L = [1,2,3,4,5,6]
with open("files/test.json", "w") as f:
    json.dump(L,f) #dump() method converts the python object to json and writes it to the file
```

```python
import json 
with open('files/test.json', 'r') as f:
    data =  json.load(f) #load() method reads the json data from the file and converts it to python object
    print(data)
    print(type(data))
```

```python
# dict
d = {
    'name':'Tawfica',
     'age':25,
     'gender':'Female'
}

with open("files/test.json", "w") as f:
    json.dump(d, f, indent=4) #indent parameter is used to format the json data in a more readable way by adding indentation to the json data
```

```python
import json 

with open("files/test.json", "r") as f:
    data = json.load(f)
    print(data)
    print(type(data))
```

# Pickling in Python

### Pickling
`Pickling` is a serialization technique of python by which python object can be converted into binary format, and `unpickling` is the inverse operation, whereby a byte stream (from a binary file or bytes-like object) is converted back into an object hierarchy.

```python
def display():
    return "Hi!my name is Tawfica and I am a Programmer"
```

```python
d = display() #storing the return value of the function in a variable
print(d)
```

```python
# pickle is a module that allows us to serialize and deserialize python objects. It converts the python object into a byte stream and vice versa. It is used to save the state of an object to a file or to send it over a network.
import pickle
with open('files/function.pkl', 'wb') as f: #.pkl is the extension of the file which is used to store the serialized object
    pickle.dump(display,f) #dump() method serializes the python object and writes it to the file but contents of the file will not be human readable as it is in binary format
```

```python
# pickle load
import pickle
with open('files/function.pkl','rb') as f: #rb is used to read the binary file
  p = pickle.load(f) #load() method deserializes the byte stream and converts it back to the original python object
  print(p)
  print(type(p))
  print(p()) #calling the function which was serialized and deserialized
```
