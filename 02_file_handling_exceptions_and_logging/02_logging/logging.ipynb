```python
# logging helps to track events that happen when some software runs. 
import logging  #built-in module in python to log events
import os       # folder creation, path manipulation, etc.
```

### Quick Breakdown of the Logging Setup

*   **Folder & File Setup:** Defines a folder named `logs` and a file named `application.log` to store all the tracking data.
*   **Safe Folder Creation:** Creates the folder safely. If the folder is already there, it quietly ignores it instead of crashing the program.
*   **Universal File Path:** Combines the folder and file name into one complete path. This ensures it works perfectly on any operating system (Windows, Mac, Linux).
*   **Core Configuration:** Sets up the main rules for how the logging system should behave:
    *   **Destination:** Tells the logger exactly where to save the file.
    *   **Layout (Format):** Designs how each log entry will look. It will show the exact time, the logger's name, the severity level, and the actual message.
    *   **Severity Level:** Sets the minimum importance level to `INFO`. The system will only record standard events, warnings, and errors, while ignoring low-level debugging noise.

```python
LOG_DIR = "logs" #create a directory named "logs" to store the log files
LOG_FILE_NAME = "application.log" # specify the name of the log file

os.makedirs(LOG_DIR, exist_ok=True) #makedirs() creates the directory if it doesn't exist, and exist_ok=True prevents an error if the directory already exists

log_path = os.path.join(LOG_DIR, LOG_FILE_NAME) #path.join() is used to create a full path to the log file by combining the directory and file name in a way that is compatible with the operating system

logging.basicConfig( #basicConfig() is used to configure the logging system. It sets up the log file, format, and logging level.
    filename=log_path, #specify the log file path
    format="[ %(asctime)s ] %(name)s - %(levelname)s - %(message)s", #format specifies how the log messages will be displayed. It includes the timestamp, logger name, log level, and the actual log message.
    level=logging.INFO #python has different logging levels (DEBUG-10, INFO-20, WARNING-30, ERROR-40, CRITICAL-50). Here, we set the logging level to INFO, which means that all messages at this level and above will be logged.
)    
```

```python
def profile(name):
    if type(name) == str:
        print(f"Welcome {name}")
    else:
        logging.info("Your input is not valid!")
```
