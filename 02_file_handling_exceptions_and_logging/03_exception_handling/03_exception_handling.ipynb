### 1. Why handle exceptions?
To prevent your program from crashing abruptly, show user-friendly error messages instead of technical jargon, and allow the code to recover and continue running safely.

### 2. How to handle (`try-except` block)
*   **`try` block:** Put the risky code here that might cause an error.
*   **`except` block:** Put the backup code here that runs only if an error occurs in the `try` block.

### 3. Different types of Exception Handling
*   **Specific Errors:** Catching exact issues to handle them individually (e.g., `except FileNotFoundError:` for missing files, or `except ZeroDivisionError:` for math errors).
*   **Generic Errors (Catch-all):** Catching any unexpected error using `except Exception as e:`, which safely captures the error message without stopping the program.

```python
try:
    with open("sample.txt10", "r") as f:
        data = print(f.read())
        print(data)

    a = 5
    b = 4
    print(a + b)


    L = [1, 2, 3]
    print(L[100])

    print(6/0)
    
except FileNotFoundError:
    print("File not found")

except NameError:
    print("Variable not defined")

except IndexError:
    print("Index out of range")

except Exception as e:
    print(e)
```

```python
#log exception into  a logger
import logging
import os

LOG_DIR = "logs"
LOG_FILE_NAME = "application.log"

os.makedirs(LOG_DIR, exist_ok = True)
log_path = os.path.join(LOG_DIR, LOG_FILE_NAME)

logging.basicConfig(
    filename = log_path,
    format = "[ %(asctime)s] %(name)s - %(levelname)s - %(message)s"
,
level = logging.INFO
)
```

# raise Exception in Python

In Python, the `raise` keyword is used to **intentionally throw or trigger an error (exception)**. 

**Why use it?**
If a specific rule or condition in your program is broken (for example, a user inputs a negative age, or tries to divide by 0), you can manually trigger an error to warn the user and stop the program from continuing with invalid data.

**Example Scenario:**
Imagine you are building a banking application where a withdrawal amount can never be negative. If a user tries to withdraw a negative amount, you can manually trigger an error like this:

`raise Exception("Withdrawal amount cannot be negative")`

This will immediately stop the program's execution and display your custom error message, preventing any further logical mistakes in the code.

```python
raise Exception("This is a custom exception")
```
