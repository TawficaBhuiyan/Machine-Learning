# 🔒 Tuples & Sets in Python


Two more collection types:
- A **tuple** is like a **list that can't be changed** — think of it as a sealed box 📦 once packed.
- A **set** is a bag of **unique** items with **no order** — like a bag of distinct marbles; duplicates just merge.

# Part A — Tuples

**Characteristics:** Ordered · **Unchangeable (immutable)** · Allows duplicates

## 1. Creating Tuples

Use parentheses `()`. ⚠️ A **single-item** tuple needs a trailing comma, or it's not a tuple!

```python
t1 = ()                 # empty tuple
print(t1, type(t1))

not_a_tuple = ("Tawfica")    # no comma -> just a string
a_tuple     = ("Tawfica",)   # comma -> real tuple
print(type(not_a_tuple))
print(type(a_tuple))
```

```python
print((1, 2, 3, 4))            # homogeneous
print((1, 2.5, True, [1, 2]))  # heterogeneous
print((1, 2, 3, (4, 5)))       # nested
print(tuple('hello'))          # from a string
```

## 2. Accessing Items (indexing & slicing) — same as lists

```python
t3 = (1, 2, 3, 4)
print(t3[0])     # first
print(t3[-2])    # second last
print(t3[2:3])   # slice
print(t3[::2])   # step
```

## 3. Tuples Are Immutable

You cannot edit or delete a single item. Both attempts raise a `TypeError`.

```python
t3 = (1, 2, 3, 4)
try:
  t3[0] = 2                 #can't reassign
except TypeError as e:
  print("AssignmentError:", e)

try:
    del t3[0]            # can't delete an item
except TypeError as e:
    print("Delete error:", e)
```

### Trick: to "edit" a tuple, convert it to a list, change it, convert back

```python
t3 = (1, 2, 3, 4)
temp = list(t3)      # tuple -> list
temp.append(200)     # edit freely
t3 = tuple(temp)     # list -> tuple
print(t3)
```

## 4. Tuple Functions & Methods

`len`, `min`, `max`, `sum`, `sorted` work on tuples. Tuple object has only two built-in  methods of its own: `count` and `index`.

```python
t = (1, 40, 2,1)
print("len   :", len(t))
print("min   :", min(t))
print("max   :", max(t))
print("sum   :", sum(t))
print("sorted:", sorted(t, reverse = True)) #sort() returns a list and changes are not in place
print(t)
```

```python
t = (1, 2, 3, 3, 40)
print("count of 3 :", t.count(3))
print("index of 3 :", t.index(3)) #returns the index of first occurence
```

## 5. Operations: `+`, `*`, membership, loop

```python
t1 = (1, 2, 3, 4)
t2 = (5, 6, 7, 8)
print(t1 + t2)   # concatenation
print(t1 * 2)    # repetition
print(1 in t1)   # membership
for i in t1:
    print(i, sep=',', end = '')
```

## 6. "Tuple Comprehension"? Use a Generator

There is **no** tuple comprehension. Putting a comprehension in `()` makes a **generator**, so wrap it in `tuple()`.

```python
print([i for i in range(1, 11)]) #python has list comprehension using []
print(tuple(i for i in range(1, 11))) #python has no tuple comprehension , so wrap generator() with tuple()
```

## 7. Tuple Unpacking

Assign each item to its own variable in one line. The count must match — unless you use `*` to grab the rest.

```python
a, b, c = (1, 2, 3) #tuple unpacking
print(a, b, c)

# Too few/many names -> ValueError
try:
    a, b, c, d = (1, 2, 3) #count not match
except ValueError as e:
    print("Unpack error:", e)
```

```python
a, b, *others = (1,2, 3, 4,5,6)   # * collects the rest into a list
print(a, b)
print(others)
```

### Zipping tuples — pair by position

```python
a = (1, 2, 3, 4)
b = (5, 6, 7, 8)
print(tuple(zip(a, b)))
```

# Part B — Sets

**Characteristics:** Unordered · Mutable · **No duplicates** · Items must be immutable (no lists/sets inside)

## 8. Creating Sets

Use `{}` with items. ⚠️ `{}` alone is an **empty dictionary**, not a set — use `set()` for an empty set.

```python
s = {}                 # empty SET (not {})
print(s, type(s))

s = set()
print(s, type(s))

s1 = {1, 2, 3}
print(s1, type(s1))
```

```python
# Order doesn't matter for equality
s1={1,2,3,4, 3}
s2 = {3,4,2,1,1}
print(s1,s2) #set makes elements ordered and unique
print({1, 2, 3, 4} == {4, 3, 2, 1})   # True

# A tuple inside is fine (immutable), a list is not
print({1, 'hello', 4.5, (1, 2, 3)})
```

```python
# A set CANNOT contain mutable items like a list or another set
try:
    bad = {1, 2, [3, 4]}
except TypeError as e:
    print("Mutable-item error:", e)
```

## 9. Accessing Items

Sets are **unordered**, so there is **no indexing or slicing**. To read items, loop over them.

```python
s1 = {1, 2, 3}
try:
    print(s1[0])          # no indexing in a set
except TypeError as e:
    print("Index error:", e)

for item in s1:           # this is how you read a set
    print(item, sep=',', end=' ')
```

## 10. Adding & Deleting Items

- `add(x)` → adds **one item**
- `update(iterable)` → adds **multiple items** from any iterable (list, tuple, set, etc.)

```python
s = {1, 2, 3, 4}
s.add(5) #add one value
print("After add one value:", s)
# s.update({11,12,13})
# s.update((11,12,13))#update iterate one by one value and add to the set
s.update([11,12,13])
s.update("Tawfica")
print("after update with more values:", s)
try:
  s.update(123)
except TypeError as e:
  print("Error is: ", e)
```

- `discard(x)` → removes an item; **no error** if missing
- `remove(x)` → removes an item; **error** if missing
- `pop()` → removes an **arbitrary item**
- `clear()` → removes **all items**

```python
s = {1, 2, 3, 4, 5}
s.discard(5)      # removes 5
s.discard(50)     # not there -> no error
print("after discard:", s)
try:
  s.remove(500) # not there -> KeyError
except KeyError as e:
    print("remove error as value is not present in the list:", e)
```

## 11. Set Math — union, intersection, difference

The real power of sets. Each has a method **and** an operator.

| Operation | Method | Operator | Meaning |
|---|---|---|---|
| Union | `union` | `\|` | everything in either |
| Intersection | `intersection` | `&` | items in **both** |
| Difference | `difference` | `-` | in first, not in second |

The `..._update` variants change the set in place.

```python
s1 = {1, 2, 3, 4}
s2 = {4, 5, 6}
print("union       :", s1 | s2)
print("intersection:", s1 & s2)
print("difference  :", s1 - s2)
```

```python
# In-place versions
s1 = {1, 2, 3, 4, 5}
s2 = {4, 5, 6, 7, 8}
s1.intersection_update(s2) #intersecting elements are upated in s1
print("s1 is now:", s1)
```

## 12. Membership, Functions & Iteration

```python
s = {3, 1, 4, 5, 2, 7}
print("len   :", len(s))
print("sum   :", sum(s))
print("min   :", min(s))
print("max   :", max(s))
print("sorted:", sorted(s))   # returns a sorted LIST
print(1 in s)
```

## ✅ Quick Recap

**Tuples** — ordered, **immutable**; index & slice like lists; only `count` & `index`; unpack with `a, b, *rest`.
**Sets** — unordered, **no duplicates**; no indexing (loop instead); `add`/`update`, `discard`/`remove`/`pop`; powerful math with `|`, `&`, `-`.
