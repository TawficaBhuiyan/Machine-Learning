## Conditional Statements in Python

A conditional statement lets a program branch: it runs one block of code
if a condition is `True`, and a different block if it's `False`. The
example below is a simple login check using `if` / `else`.

````python
# Login function demo - inceptionbd.com

"""
Actual Credentials:

Email: bhuiyantawfica@gmail.com
Password: 1234_haha_1234
"""

email = input("Enter your email: ")
password = input("Enter your password: ")

actual_email = "bhuiyantawfica@gmail.com"
actual_password = "1234_haha_1234"


if email == actual_email and password == actual_password:
  print("Login Successful")


else:
  print("Invalid Credentials!")
````

### Multiple Conditions (`elif`)

Add `elif` ("else if") to handle more than two possible outcomes — here
we tell the user *why* the login failed: wrong email, or the right email
with the wrong password. (Uses the same credentials shown above.)

````python
# Login function demo - inceptionbd.com

email = input("Enter your email: ")
password = input("Enter your password: ")

actual_email = "bhuiyantawfica@gmail.com"
actual_password = "1234_haha_1234"


if email == actual_email and password == actual_password:
  print("Login Successful")


elif email == actual_email and password != actual_password:
  print("Incorrect Password")


else:
  print("Invalid Credentials!")
````

### Nested Conditional Statements

An `if` block can contain another `if` / `else` inside it. Here, if the
password is wrong, the user gets a second chance to re-enter it before
the program decides success or failure. (Uses the same credentials shown
above.)

````python
# Login function demo - inceptionbd.com

email = input("Enter your email: ")
password = input("Enter your password: ")

actual_email = "bhuiyantawfica@gmail.com"
actual_password = "1234_haha_1234"


if email == actual_email and password == actual_password:
  print("Login Successful")


elif email == actual_email and password != actual_password:
  print("Incorrect Password, Please try again!")
  password = input("Enter your password again: ")

  if password == actual_password:
    print("Success!")

  else:
    print("Wrong!")


else:
  print("Invalid Credentials!")
````

### `if` / `elif` / `else` Syntax Template

The general shape of a conditional statement looks like this (this is a
template to illustrate the structure, not runnable code):

```python
if condition:
    # write something
elif condition:
    # write something
else:
    # write something else
```

### Builtins Module / Package

These ship with Python itself — no installation needed, just `import`
them:
- `math`
- `keyword`
- `random`
- `datetime`

````python
import math

print(math.sqrt(200))       # square root
print(math.factorial(4))    # 4! = 4 * 3 * 2 * 1
````

````python
import keyword

print(keyword.kwlist)   # every reserved word in Python
````

````python
import random

print(random.randint(1, 100))   # random integer 1-100 (inclusive) — changes every run
````

````python
import datetime

print(datetime.datetime.now())   # the current date and time
````
