# Loops in Python — Doing Things Again and Again, the Smart Way 🔁

Imagine you needed to print `"Hello"` 100 times. Would you write
`print("Hello")` on 100 separate lines? Of course not — that's exactly the
kind of repetitive job a **loop** is built for.

A loop runs the same block of code over and over, as many times as you
need, without copy-pasting it. Python gives you two loop keywords: `for`
and `while`. This notebook walks through both, one small idea at a time —
try predicting each output before you run the cell.

## The `for` Loop

A `for` loop walks through a sequence of items, one at a time, and runs
the same block of code for each one:

```python
for variable in sequence:
    # do something with variable and here 'in' is a membership  operator
```

The simplest sequence to loop over is one generated by `range()`.

````python
# range(1, 10) generates the numbers 1 through 9 -- 10 is NOT included
for i in range(1, 10):
    print(i,"Times")
````

### Controlling `range()`

`range(start, stop)` counts up from `start` up to (but not including)
`stop`. Add a third number, `range(start, stop, step)`, to control the
size of each jump — even backwards, with a negative step.

````python
for i in range(1, 11):        # default step size 1
    print(i)
````

````python
for i in range(11, 0, -1):    # counting down from 11 to 1
    print(i)
print("\n")
for i in range(1, 11, 2):     # every 2nd number: 1, 3, 5, 7, 9
    print(i)
````

### Looping Over Other Things — Not Just Numbers

`for` doesn't only work with `range()` — it can walk through *any*
sequence: the characters of a string, the items of a list, or the items
of a tuple. And the loop variable can be named anything you like, not
just `i`.

````python
for letter in "Tawfica":    # loops over each character in the string
  print(letter, end = '')
print("\n")
for number in [1, 2, 3, 4]:   # loops over a list
    print(number, end = ',')
print("\n")
for number in (1, 2, 3, 4):   # loops over a tuple -- same idea, different container
    print(number, end = " ")
````

### A Common Mistake: Looping Over a Plain Number

`for` needs something *iterable* — a sequence it can step through. A
single number like `1.2` isn't iterable, so this raises a `TypeError`:

````python
for i in 1.2:
    print(i)
````

## Nested Loops

A loop can contain another loop inside it — this is called **nesting**.
For every single step of the outer loop, the entire inner loop runs from
start to finish. This is handy for grids, tables, and pairs of values.

````python
for i in range(1, 5):
    for j in range(1, 5):
        print(i, j)
````

## Loop Control Statements

Sometimes you want to change a loop's normal flow:

| Statement | What it does |
|---|---|
| `break` | Stops the loop immediately, no matter how many rounds were left |
| `continue` | Skips the rest of the current round and jumps to the next one |
| `pass` | Does nothing — a placeholder so an empty block doesn't cause an error |

````python
# break -- stop the loop as soon as we find what we're looking for
for i in range(1, 11):
    print(i)
    if i == 5:
        break
````

`break` is especially handy for searching — once you've found the
item, there's no need to keep looking:

````python
names = ["tawfica","bhuiyan",1,2,3,"SR","TB"]

for name in names:
    if name == "SR":
        print(name, "found!")
        break
````

````python
# continue -- skip just this one round, then carry on
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
````

````python
# pass -- a no-op placeholder, useful when a block needs *something*
# written but you don't want it to do anything yet
for i in range(1, 11):
    pass

print("Ok")   # the loop above ran 10 times and did nothing each time
````

## The `while` Loop

Where `for` repeats a fixed number of times (or once per item in a
sequence), `while` repeats **as long as a condition stays `True`**. You
don't always know in advance how many times it will run.

```python
while condition:
    # code that (eventually) makes condition False
```

````python
i = 1
while i <= 5:
 print(i,"Tawfica")
 i += 1      # without this line, the condition never becomes False!
````

### ⚠️ Watch Out for Infinite Loops

If the condition never becomes `False`, the loop never stops. `while
True:` is *always* true, so the cell below runs forever until it's
manually interrupted (in Colab, click the ■ Stop button next to the
cell; in Jupyter, use "Interrupt Kernel"):

````python
while True:
    print("Tawfica")
````

### `while` with `else`

A `while` loop can have an `else` block, which runs once — automatically
— right after the loop finishes normally (as long as it wasn't stopped
early with `break`).

````python
x = 1
while x < 4:
  print(x)
  x += 1

else:
    print("Program Executed")
````

## Putting It Together: A Multiplication Table

Let's use a `while` loop to build a small, practical program: given any
number, print its multiplication table from 1 to 10. For example, if the
user enters `2`, the target output looks like this:

```
2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
...
2 x 10 = 20
```

````python
number = int(input("Enter the number: ")) #need type conversion, as input returns string type
i = 1
while i <= 10:
  print(number, ' * ', i , '  = ', number * i)
  i += 1
````

## Recap

- `for` loops step through a sequence — use `range()` for a fixed count,
  or loop directly over strings, lists, and tuples.
- `while` loops repeat based on a condition — make sure something inside
  the loop eventually makes that condition `False`.
- `break`, `continue`, and `pass` give you fine control over what happens
  inside a loop.

The fastest way to get comfortable with loops is to experiment — go back
up, change the numbers, and re-run the cells to see what changes.
