# 📋 Lists in Python

A **list** stores many items under **one name**. Think of a **shopping list** 🛒 — one page, many items, and you can add, remove, or reorder them any time.

**This notebook covers**
- What a list is & its characteristics
- Creating lists
- Accessing items (indexing & slicing)
- Editing, adding & deleting items
- Built-in functions & list methods
- Copying (and the aliasing trap)
- Operations, list comprehension & `zip()`

> ▶️ Run each cell (**Shift + Enter**). The one intentional error is wrapped in `try/except`, so the whole notebook runs top-to-bottom without crashing.

## 1. What is a List?

A list holds multiple values in a single variable. Technically it behaves like a **dynamic array** — it can grow or shrink on the fly.

```python
L = [1, 2, 3]
print(L)
print(type(L))
```

### Characteristics of a list
- **Ordered** — items keep their position
- **Mutable** — you can change it after creation
- **Heterogeneous** — can mix types (int, str, float, even other lists)
- **Allows duplicates**
- **Dynamic** — grows/shrinks
- **Nestable** — a list can contain other lists

```python
L1 = [1, 2, 3]
L2 = [2, 3, 1]
print(L1 == L2)      # False -> order matters
print([1, 2, 3, 3])  # duplicates are allowed
```

## 2. Creating Lists

```python
print([])                                # empty
print([1, 2, 3, 4, 5])                   # 1D
print([1, 2, 3, [4, 5]])                 # 2D (nested)
print([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])  # 3D
print([1, True, 5.6, 5 + 6j, 'Hello'])   # heterogeneous
print(list("Tawfica Bhuiyan"))                     # from a string
```

## 3. Accessing Items

- **Indexing** starts at `0`; negative index counts from the end.
- **Nested** lists need one index per level.

```python
l = [1, 2, 3, 4, 5]
print(l[0])    # first item
print(l[-1])   # last item
```

```python
nested = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]
print(nested[0][1][1])   # -> 4  (go layer by layer)
print(nested[1][1][0])   # -> 7
```

### Slicing — grab a piece: `l[start:stop:step]`
`start` is included, `stop` is **not**.

```python
l = [1, 2, 3, 4, 5]
print(l[0:2])   # [1, 2]
print(l[:])     # whole copy
print(l[::3])   # every 3rd item
```

## 4. Lists Are Mutable — Editing Items

Unlike strings, you **can** change a list in place, by index or by slice.

```python
# Reminder: strings are immutable, lists are not.
s = "Tawfica"
try:
    s[0] = "A"            # strings can't be changed
except TypeError as e:
    print("String error:", e)

L = [1, 2, 3, 4]
L[0] = 100                # lists can be changed
print("List after edit:", L)
```

```python
# Editing with a slice (replace a range at once)
L = [1, 2, 3, 4]
L[1:4] = [10, 100]
print(L)
```

## 5. Adding Items

- `append(x)` → add **one** item at the end
- `extend(iterable)` → add **each** item of another list
- `insert(index, value)` → add the value at index  position

```python
L = [1, 2, 3, 4]
L.append(10)       # one item at the end
print("append:", L)

L.append([1,2, 3, 4])  # a list added as a SINGLE item
print("append list:", L)
```

```python
L = [1, 2, 3, 4]
l = [10, 20, 30, 40]
L.extend(l)   # merges the items in
print("extend:", L)

L.insert(1, 20)        # insert 200 at index 1
print("insert:", L)
```

## 6. Deleting Items

- `del L[i]` → delete by index (or a slice)
- `remove(x)` → delete by **value**
- `pop()` → remove & return the last item
- `clear()` → empty the whole list

```python
L = [1, 2, 3, 4]
del L[0]            #delete one index
print("After 0 index deletion:", L)

L = [1, 2, 3, 4]
del L[1:3]          #delete a range
print("del slice:", L)
```

```python
L = [4, 1, 2, 3, 4]
L.remove(4)          # delete the first occurence
print("Delete the first occurence of any value using remove:", L)

L = [1, 2, 3, 4]
deleted_last_item = L.pop()              # delete the last item and return its value
print(f"After deleting the last item {deleted_last_item} using pop :", L)

L.clear()            # now empty
print("clear:", L)
```

## 7. Built-in Functions & Methods

`len`, `min`, `max`, `sorted` work on any list. Methods `count`, `index`, `reverse`, `sort` belong to the list itself.

```python
L = [1, 2, 3, 4]
print("len   :", len(L))
print("min   :", min(L))
print("max   :", max(L))
print("sorted:", sorted(L))
```

```python
L = [1, 2, 3, 4, 4, 4, 5, 5]
print("count of 4 :", L.count(4))   # how many 4s
print("index of 5 :", L.index(5))   # first position of 5
```

```python
L = [4, 2, 1, 55, 10]
L.reverse()          # reverse in place
print("reversed:", L)

print(sorted(L, reverse = True)) # changes not in L , just used to print
print(L)

L.sort()             # sort in place (ascending)
print("sorted  :", L)

L.sort(reverse = True) #sort in descending order in  place
print("Reversed:", L)
```

## 8. Copying — and the Aliasing Trap ⚠️

`b = a` does **not** make a new list — both names point to the **same** list. Use `.copy()` for an independent copy.`id()` refers to the object identity in memory

```python
L = [1, 2, 3, 4]
l1 = L.copy()        # independent copy
L.append(2000)       # change the original only
print("original:", L)
print("copy    :", l1)   # unaffected
print(id(L1)) #L1 refers to which object identity
print("same object?", "yes" if(id(L) == id(l1)) else "No")   # False
```

## 9. Operations on Lists

`+` joins, `*` repeats, `in` checks membership, and a `for` loop walks each item.

```python
L1 = [1, 2, 3, 4, 5]
L2 = [5, 6, 7, 8, 9]
print(L1 + L2)   # concatenation
print(L1 * 2)    # repetition(l1 repeats 2 times)
```

```python
#membership operator
L1 = [1, 2, 3, 4, 5]
print(3 in L1)       # True
print(40 in L1)      # False

for item in L1:
    print(item, sep = ',', end =' ')
```

## 10. List Comprehension

A short, readable way to build a list in one line:
`[expression for item in iterable if condition]`

```python
# Long way: numbers 1..10
L = []
for i in range(1, 11):
  L.append(i)
print("Created by Loop: ", L)

# Short way (comprehension)
LC = [i for i in range(1, 11)] #list comprehension
print("Created in comprehensive way: ",LC)
```

```python
# With a condition: numbers divisible by 5 in 1..50
LC = [i for i in range(1,21) if i%5 == 0]
print(LC)
```

## 11. `zip()` — Pair Up Two Lists

`zip()` pairs items by position. If lengths differ, the **shortest** one decides the result length.

```python
roll_number  = [1, 2, 3, 4, 5]
student_name = ['Tawfica', 'Alex', 'Harish', 'Kumer', 'Bhuiyan','Mony']
print(zip(roll_number, student_name)) #returns an iterator
print(list(zip(roll_number, student_name)))
```

## ✅ Quick Recap

- A list = ordered, **mutable** collection; index with `L[i]`, slice with `L[a:b:c]`.
- **Add:** `append`, `extend`, `insert`. **Delete:** `del`, `remove`, `pop`, `clear`.
- **Info:** `len`, `min`, `max`, `sorted`, `count`, `index`; **in place:** `reverse`, `sort`.
- Copy with `.copy()` — plain `=` only makes an alias.
- **Comprehensions** build lists in one line; **`zip`** pairs lists together.
