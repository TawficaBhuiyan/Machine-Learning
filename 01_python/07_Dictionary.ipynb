# 🗂️ Dictionaries in Python


A **dictionary** stores data as **key → value** pairs, like a real dictionary maps a *word* to its *meaning* 📖, or a contact list maps a *name* to a *number*.

```python
person = {'name': 'Tawfica Bhuiyan', 'age': 22, 'gender': 'female'}
```

**Characteristics**
- **Mutable** (can be changed)
- Accessed by **key**, not by position (no indexing)
- Keys must be **unique** and **immutable** (str, number, tuple — not a list)

> ▶️ Run each cell in order. Intentional errors are wrapped in `try/except`, so the notebook runs top-to-bottom without crashing.

## 1. Creating Dictionaries

````python
d = {}                    # empty dictionary
print(d, type(d))

d1 = {"name": "Tawfica", "age": 22, "gender": "female"}   # 1D
print(d1)
````

````python
# A tuple key is allowed (immutable); a list key is NOT
d2 = {(1, 2, 3): 33, "name": "Tawfica"}
print(d2)

try:
    bad = {[1, 2, 3]: 33}      # list can't be a key
except TypeError as e:
    print("Key error:", e)
````

````python
# Nested dictionary (a value can be another dictionary)
student = {
    "name": "Tawfica",
    "college_name": "ABC",
    "semester": 5,
    "subjects": {"DSA": 80, "Math": 50, "English": 70}
}
print(student)

# From a list of pairs, using dict()
d4 = dict([("name", "Tawfica"), ("age", 22)])
print(d4)
````

````python
# Duplicate keys: the LAST value wins
d5 = {"name": "Tawfica", "name": "Bhuiyan", "age": 22}
print(d5)
````

## 2. Accessing Items

Read a value by its **key** with `d[key]`. There is no number indexing — a missing key raises a `KeyError`.

````python
d1 = {"name": "Tawfica", "age": 22, "gender": "female"}
print(d1["name"])
print(d1["age"])

try:
    print(d1[0])          # 0 is not a key here
except KeyError as e:
    print("KeyError:", e)
````

````python
# Nested access -> one key per level
student = {"name": "Tawfica", "subjects": {"DSA": 80, "Math": 50}}
print(student["subjects"]["Math"])
````

## 3. Adding & Editing

Assigning to a **new** key adds it; assigning to an **existing** key updates it.

````python
d1 = {"name": "Tawfica", "age": 22}
d1["weight"] = 70          # new key -> added
print("added :", d1)

d1["name"] = "Bhuiyan"     # existing key -> updated
print("edited:", d1)
````

## 4. Removing Items

- `pop(key)` → remove by key, returns its value, if key not exist -> KeyError
- `popitem()` → remove the last inserted pair
- `del d[key]` → delete by key, returns nothing, it key not exist -> KeyError
- `clear()` → empty the dictionary

````python
d1 = {"name": "Tawfica", "age": 22, "gender": "female", "Address": "Comilla"}
popped_value = d1.pop('gender')
print(f"Popped values: {popped_value} \npopped dictionary: {d1}")

d1 = {"name": "Tawfica", "age": 22, "gender": "female"}
d1.popitem()
print("popitem :", d1)

del d1["name"]
print("del :", d1)

d1.clear()
print("clear   :", d1)
````

## 5. Operations: membership & iteration

`in` checks the **keys**. Looping also walks the keys, so `d[key]` gives each value.

````python
d = {"name": "Tawfica", "age": 22, "gender": "female"}
print("name" in d)       # True
print("email" in d)      # False

#iterating through keys
for key in d:
  print(f"'{key}' : {d[key]}", end = ', ')
````

## 6. Functions & Methods

`len`, `sorted`, `min`, `max` operate on the **keys**. The trio `items()`, `keys()`, `values()` is used constantly.

````python
d = {"name": "Tawfica", "age": 22, "gender": "female"}
print("len   :", len(d))
print("sorted:", sorted(d))   # sorts the keys
````

````python
d = {"name": "Tawfica", "age": 22, "gender": "female"}
print(d.items())    # shows key-value pairs
print(d.keys())     # shows keys only
print(d.values())   # shows values only
````

````python
# update: merge another dictionary in (matching keys get overwritten)
d1 = {1: 2, 3: 4, 4: 5}
d2 = {4: 7, 6: 8, 4: 10}
d1.update(d2)
print(d1)
````

## 7. Dictionary Comprehension

Build a dictionary in one line: `{key: value for item in iterable}`.

````python
# First 10 numbers mapped to their squares
squares = {i: i**2 for i in range(1, 11)}
print(squares)
l1 = [i for i in range(1, 11)]
l2 = [i for i in range(1, 31) if i % 2 == 0]
print(l1, l2)
squares = {i: j**2 for i, j in zip(l1, l2) }
print(squares)
````

## ✅ Quick Recap

- A dictionary = **key → value** pairs; read with `d[key]` (no number indexing).
- Keys are **unique** & **immutable**; duplicate keys keep the **last** value.
- **Add/edit:** `d[key] = value`. **Remove:** `pop`, `popitem`, `del`, `clear`.
- **Views:** `items()`, `keys()`, `values()`; merge with `update()`.
- Build fast with **dict comprehension**: `{k: v for ...}`.
